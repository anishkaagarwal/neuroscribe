```python
import os
from groq import Groq

# Set GROQ_API_KEY in your environment (or a .env file) - never hard-code it.
client = Groq(api_key=os.environ["GROQ_API_KEY"])


completion = client.chat.completions.create(
    model="openai/gpt-oss-20b",   # or try a smaller one if needed
    messages=[
        {"role": "system", "content": "You are a helpful assistant that summarizes conversations."},
        {"role": "user", "content": f"Summarize the following conversation in a concise abstract(smaller that Summary just the gist) and a short summary:\n\n{conversation}"}
    ],
    temperature=0.7,
    max_completion_tokens=512,   # reduce from 4096 to avoid very long outputs
    top_p=1,
    reasoning_effort="low",
    stream=True,
)

for chunk in completion:
    if chunk.choices[0].delta.content:
        print(chunk.choices[0].delta.content, end="")
```
